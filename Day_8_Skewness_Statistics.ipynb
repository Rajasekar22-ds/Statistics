## 1. What is Skewness?

**Skewness measures how much a distribution is asymmetric.**

In simple words:

> Skewness tells us whether the data has a longer tail on the left or right side.

There are three common situations:

| Type | Meaning |
|---|---|
| **Symmetric** | Data is roughly balanced on both sides |
| **Positive skew / Right skew** | Long tail extends toward larger values |
| **Negative skew / Left skew** | Long tail extends toward smaller values |

### Easy memory trick

- **Positive skew → tail points RIGHT ➡️**
- **Negative skew → tail points LEFT ⬅️**



## 2. Symmetric Distribution

In a roughly symmetric distribution:

**Mean ≈ Median**

Example:

`10, 20, 30, 40, 50`

The values are balanced around the center.

A histogram of a symmetric distribution tends to look similar on both sides of its center.


````python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import skew
````

````python

data = [10, 20, 30, 40, 50]

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Skewness:", skew(data))

plt.hist(data, bins=5, edgecolor="black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Approximately Symmetric Data")
plt.show()
````


## 3. Positive Skewness (Right-Skewed)

A distribution is **positively skewed** when its tail extends toward larger values.

Example:

`10, 11, 12, 13, 14, 15, 50`

The value `50` pulls the mean toward the right.

Usually:

**Mean > Median**

Think about income:

Most people may have moderate incomes, while a small number of people have extremely high incomes.


````python

positive_data = [10, 11, 12, 13, 14, 15, 50]

print("Mean:", np.mean(positive_data))
print("Median:", np.median(positive_data))
print("Skewness:", skew(positive_data))

plt.hist(positive_data, bins=7, edgecolor="black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Positive / Right Skewed Data")
plt.show()
````


## 4. Negative Skewness (Left-Skewed)

A distribution is **negatively skewed** when its tail extends toward smaller values.

Example:

`10, 45, 46, 47, 48, 49, 50`

The low value `10` pulls the mean toward the left.

Usually:

**Mean < Median**

A common example can be an exam where most students score very high but a small number score very low.


````python

negative_data = [10, 45, 46, 47, 48, 49, 50]

print("Mean:", np.mean(negative_data))
print("Median:", np.median(negative_data))
print("Skewness:", skew(negative_data))

plt.hist(negative_data, bins=7, edgecolor="black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Negative / Left Skewed Data")
plt.show()
````


## 5. How to Interpret the Skewness Value

A common practical interpretation is:

| Skewness | General interpretation |
|---:|---|
| Around 0 | Approximately symmetric |
| Greater than 0 | Positive / right skew |
| Less than 0 | Negative / left skew |

For example:

- `0.05` → almost symmetric
- `1.20` → positively skewed
- `-1.10` → negatively skewed

⚠️ These are **rules of thumb**, not strict universal cutoffs. The appropriate interpretation depends on the dataset and context.



## 6. Mean and Median Relationship

A useful beginner rule:

### Right skew
**Mean > Median**

### Left skew
**Mean < Median**

### Symmetric
**Mean ≈ Median**

But remember: this is a useful clue, not a mathematical rule that works perfectly for every dataset.


````python

examples = {
    "Symmetric": [10, 20, 30, 40, 50],
    "Right Skewed": [10, 11, 12, 13, 14, 15, 50],
    "Left Skewed": [10, 45, 46, 47, 48, 49, 50]
}

for name, values in examples.items():
    print("\n", name)
    print("Mean   :", np.mean(values))
    print("Median :", np.median(values))
    print("Skewness:", skew(values))
````


## 7. Skewness Using SciPy

SciPy provides a convenient function:

```python
from scipy.stats import skew

skew(data)
```

By default, SciPy calculates a bias-corrected form only when requested with `bias=False`.

For learning and practical analysis, always check the function's documentation when the exact estimator matters.


````python

data = [10, 11, 12, 13, 14, 15, 50]

print("Default SciPy skewness:", skew(data))
print("Bias-corrected skewness:", skew(data, bias=False))
````


## 8. Skewness Using Pandas

For a Pandas Series, we can use:

```python
series.skew()
```

Pandas' `.skew()` is convenient when working with DataFrames.


````python

df = pd.DataFrame({
    "Salary": [25000, 28000, 30000, 32000, 35000, 40000, 150000],
    "Age": [22, 23, 24, 25, 26, 27, 28]
})

print(df)

print("\nSkewness:")
print(df.skew(numeric_only=True))
````


## 9. Comparing Different Distributions

Let's create three datasets and compare their skewness.

This helps us see that skewness is not just a number — it describes the shape of the distribution.


````python

symmetric = np.random.normal(loc=50, scale=10, size=1000)
right_skewed = np.random.exponential(scale=10, size=1000)
left_skewed = -np.random.exponential(scale=10, size=1000) + 50

print("Symmetric skewness :", skew(symmetric))
print("Right skewness     :", skew(right_skewed))
print("Left skewness      :", skew(left_skewed))
````

````python

plt.hist(symmetric, bins=30, edgecolor="black")
plt.title("Approximately Symmetric Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

plt.hist(right_skewed, bins=30, edgecolor="black")
plt.title("Right-Skewed Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

plt.hist(left_skewed, bins=30, edgecolor="black")
plt.title("Left-Skewed Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
````


## 10. Skewness and Outliers

Outliers can strongly affect skewness.

For example:

`20, 21, 22, 23, 24, 25`

has little reason to be strongly skewed.

But adding:

`200`

can create strong right skewness.

This is why we should inspect the data and not blindly trust a single statistic.


````python

normal_values = [20, 21, 22, 23, 24, 25]
outlier_values = [20, 21, 22, 23, 24, 25, 200]

print("Without outlier:", skew(normal_values))
print("With outlier   :", skew(outlier_values))

plt.boxplot(outlier_values)
plt.title("Effect of an Outlier")
plt.ylabel("Value")
plt.show()
````


## 11. Real-World Example — House Prices

Suppose house prices are:

`₹20L, ₹22L, ₹25L, ₹27L, ₹30L, ₹35L, ₹150L`

Most houses are in a lower range, but one very expensive house creates a long right tail.

This type of data can be **positively skewed**.

For heavily skewed variables, analysts may consider transformations such as a logarithm before certain statistical or machine-learning tasks.


````python

house_prices = np.array([20, 22, 25, 27, 30, 35, 150])

print("Mean price:", house_prices.mean())
print("Median price:", np.median(house_prices))
print("Skewness:", skew(house_prices))
````


## 12. Log Transformation Example

A logarithmic transformation can reduce strong right skewness.

This does **not** mean we should always transform the data.

Whether to transform depends on the analysis, model, and interpretation requirements.


````python

income = np.array([20000, 22000, 25000, 30000, 35000, 50000, 200000])

original_skew = skew(income)
log_income = np.log(income)
log_skew = skew(log_income)

print("Original skewness:", original_skew)
print("Log-transformed skewness:", log_skew)

plt.hist(income, bins=7, edgecolor="black")
plt.title("Original Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.show()

plt.hist(log_income, bins=7, edgecolor="black")
plt.title("Log-Transformed Income Distribution")
plt.xlabel("Log Income")
plt.ylabel("Frequency")
plt.show()
````


## 13. Skewness vs Variance vs Standard Deviation

These statistics answer different questions:

| Statistic | Main question |
|---|---|
| **Mean** | Where is the center? |
| **Median** | What is the middle value? |
| **Variance** | How spread out is the data? |
| **Standard Deviation** | How much does data typically vary around the mean? |
| **Skewness** | Which direction is the distribution asymmetric? |

So:

**Spread ≠ Skewness**

A dataset can have high variability without being strongly skewed.

## 15. Practical Data Science Workflow

When analyzing a numeric feature:

1. Calculate the mean.
2. Calculate the median.
3. Calculate skewness.
4. Plot a histogram.
5. Check a boxplot.
6. Investigate unusual values.
7. Decide whether transformation is appropriate.
8. Document the reason for your decision.

Do not remove or transform data simply because it is skewed. First understand what the values represent.
