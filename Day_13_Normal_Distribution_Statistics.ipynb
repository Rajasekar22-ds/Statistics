# Day 13 — Normal Distribution

A **normal distribution** is a symmetric, bell-shaped distribution. Many real-world measurements can be approximately normal.

In this notebook:
- Bell curve, mean, and standard deviation
- 68–95–99.7 rule
- Z-scores
- Standard normal distribution
- Probabilities and percentiles with SciPy
- Python plots and practice exercises

## 1. Features of a normal distribution

- The curve is symmetric around the mean.
- Mean, median, and mode are equal for an ideal normal distribution.
- Values near the mean are more common than values far away.
- The total area under the curve is 1.

Not all real-world data is normal, so inspect your data before assuming this model.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 500)
pdf = (1 / np.sqrt(2 * np.pi)) * np.exp(-0.5 * x**2)

plt.plot(x, pdf)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Standard Normal Distribution")
plt.show()
```

## 2. Mean and standard deviation

- **Mean (μ):** the center of the curve.
- **Standard deviation (σ):** how spread out the values are.

A smaller standard deviation makes a narrower curve; a larger one makes a wider curve.

```python
x = np.linspace(-10, 10, 1000)

for mean, std in [(0, 1), (0, 2), (3, 1)]:
    y = (1 / (std * np.sqrt(2*np.pi))) * np.exp(-0.5*((x-mean)/std)**2)
    plt.plot(x, y, label=f"mean={mean}, SD={std}")

plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Effect of Mean and Standard Deviation")
plt.legend()
plt.show()
```

## 3. The 68–95–99.7 rule

For an approximately normal distribution:
- About **68%** of values lie within 1 standard deviation of the mean.
- About **95%** lie within 2 standard deviations.
- About **99.7%** lie within 3 standard deviations.

Example: mean exam score = 70 and SD = 10.
- 68%: approximately 60 to 80
- 95%: approximately 50 to 90
- 99.7%: approximately 40 to 100

```python
mean = 70
std = 10

print("About 68%:", mean-std, "to", mean+std)
print("About 95%:", mean-2*std, "to", mean+2*std)
print("About 99.7%:", mean-3*std, "to", mean+3*std)
```

## 4. Z-score

A **Z-score** tells how many standard deviations a value is above or below the mean.

**z = (x − μ) / σ**

- z = 0: at the mean
- z = 1: one SD above the mean
- z = −2: two SD below the mean

Example: score = 85, mean = 70, SD = 10. Z = (85 − 70)/10 = 1.5.

```python
score = 85
mean = 70
std = 10

z_score = (score - mean) / std
print("Z-score:", z_score)
```

## 5. Standard normal distribution

The standard normal distribution has mean 0 and standard deviation 1. A Z-score converts a value to this scale.

```python
from scipy.stats import norm

z = np.linspace(-4, 4, 500)
plt.plot(z, norm.pdf(z))
plt.axvline(0, linestyle="--")
plt.xlabel("Z-score")
plt.ylabel("Density")
plt.title("Standard Normal Curve")
plt.show()
```

## 6. Probability with SciPy

- `norm.cdf(x)` gives the probability below or equal to x.
- `1 - norm.cdf(x)` gives the probability above x.
- `norm.pdf(x)` gives the density at x, not the probability at an exact point.
- `norm.ppf(p)` finds the value at cumulative probability p.

For a general normal distribution, use `loc=mean` and `scale=standard_deviation`.

```python
# Standard normal examples
print("P(Z <= 1):", norm.cdf(1))
print("P(Z > 1):", 1 - norm.cdf(1))
print("P(-1 <= Z <= 1):", norm.cdf(1) - norm.cdf(-1))
```

## 7. Example: exam scores

Suppose scores are approximately normal with mean 70 and SD 10.

Find the probability of a score below 85.

```python
mean = 70
std = 10
score = 85

z = (score - mean) / std
p_below = norm.cdf(score, loc=mean, scale=std)

print("Z-score:", z)
print("P(Score <= 85):", p_below)
print("Percentage:", p_below * 100, "%")
```

## 8. Probability between two values

**P(a < X < b) = CDF(b) − CDF(a)**

Example: probability that a score lies between 60 and 80, with mean 70 and SD 10.

```python
p_between = norm.cdf(80, loc=70, scale=10) - norm.cdf(60, loc=70, scale=10)
print("P(60 < Score < 80):", p_between)
print("Percentage:", p_between * 100, "%")
```

## 9. Find a percentile

The 95th percentile is the value below which 95% of observations lie, under the model.

```python
percentile_95 = norm.ppf(0.95, loc=70, scale=10)
print("95th percentile:", percentile_95)
```

## 10. Plot probability area

The area under a PDF between two values represents the probability in that interval.

```python
mean, std = 70, 10
x = np.linspace(35, 105, 600)
y = norm.pdf(x, loc=mean, scale=std)

plt.plot(x, y)
plt.fill_between(x, y, where=(x >= 60) & (x <= 80), alpha=0.3)
plt.xlabel("Exam score")
plt.ylabel("Density")
plt.title("Area for Scores Between 60 and 80")
plt.show()
```

## 11. Inspect data with a histogram

A bell-shaped histogram may be consistent with normality, but a histogram alone does not prove data is normally distributed.

```python
np.random.seed(42)
sample = np.random.normal(loc=70, scale=10, size=1000)

plt.hist(sample, bins=30, edgecolor="black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Simulated Normal Data")
plt.show()

print("Sample mean:", sample.mean())
print("Sample standard deviation:", sample.std(ddof=1))
```

## Common mistakes

1. A Z-score is not a probability; it measures distance from the mean in SD units.
2. PDF height is not an exact-value probability; continuous probabilities are areas over intervals.
3. The 68–95–99.7 rule is approximate and assumes a normal model.
4. Not all datasets are normal.
5. `norm.cdf(x)` gives probability below x; use `1 - norm.cdf(x)` for above x.

## Practice exercises

1. If mean = 100 and SD = 15, find the Z-score for x = 130.
2. Calculate P(Z ≤ 0).
3. For scores with mean 70 and SD 10, calculate P(Score > 80).
4. Find the 90th percentile for a normal distribution with mean 50 and SD 8.
5. Generate 1,000 normal values and plot a histogram.

```python
# Practice answers / examples
print("1. Z-score:", (130-100)/15)
print("2. P(Z <= 0):", norm.cdf(0))
print("3. P(Score > 80):", 1 - norm.cdf(80, loc=70, scale=10))
print("4. 90th percentile:", norm.ppf(0.90, loc=50, scale=8))

sample = np.random.normal(loc=50, scale=8, size=1000)
plt.hist(sample, bins=30, edgecolor="black")
plt.title("1,000 Simulated Normal Values")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```
