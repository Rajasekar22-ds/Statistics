# Day 4 — Range (Statistics)

A beginner-friendly Jupyter Notebook covering **Range** using Python, NumPy, and Pandas.

## 1. What is Range?

**Range = Maximum Value − Minimum Value**

Range tells us how spread out the values are.

````python
data = [10, 20, 30, 40, 50]
maximum = max(data)
minimum = min(data)
print("Maximum:", maximum)
print("Minimum:", minimum)
print("Range:", maximum - minimum)
````

## 2. Manual Calculation

````python
data = [45, 20, 70, 10, 55]
sorted_data = sorted(data)
print("Sorted Data:", sorted_data)
print("Range:", sorted_data[-1] - sorted_data[0])
````

## 3. Custom Range Function

````python
def calculate_range(data):
    return max(data) - min(data)

values = [10, 25, 35, 50, 70]
print("Range:", calculate_range(values))
````

## 4. Range Using NumPy

````python
import numpy as np

data = np.array([10, 20, 30, 40, 50])
print("Maximum:", np.max(data))
print("Minimum:", np.min(data))
print("Range:", np.max(data) - np.min(data))
print("Range using np.ptp():", np.ptp(data))
````

## 5. Student Marks Example

````python
marks = np.array([78, 65, 82, 90, 55, 72, 88, 69, 95, 74])
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Range:", np.ptp(marks))
````

## 6. Row-wise and Column-wise Range

````python
marks = np.array([
    [80, 70, 90],
    [60, 75, 85],
    [90, 88, 95],
    [70, 65, 72],
    [85, 80, 88]
])

print("Overall Range:", np.ptp(marks))
print("Column-wise Range:", np.max(marks, axis=0) - np.min(marks, axis=0))
print("Row-wise Range:", np.max(marks, axis=1) - np.min(marks, axis=1))
````

## 7. Range Using Pandas

````python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Kumar", "Arun", "Priya", "Divya"],
    "Marks": [80, 70, 90, 65, 85]
})

print(df)
print("Range:", df["Marks"].max() - df["Marks"].min())
````

## 8. Range of Multiple Columns

````python
df = pd.DataFrame({
    "Maths": [80, 70, 90, 65, 85],
    "Science": [75, 80, 95, 70, 88],
    "English": [85, 72, 88, 75, 90]
})

print(df.max() - df.min())
````

## 9. Group-wise Range

````python
df = pd.DataFrame({
    "Department": ["IT", "IT", "IT", "HR", "HR", "HR", "Sales", "Sales", "Sales"],
    "Salary": [30000, 35000, 45000, 25000, 28000, 30000, 32000, 36000, 40000]
})

result = df.groupby("Department")["Salary"].agg(lambda x: x.max() - x.min())
print(result)
````

## 10. Range With Missing Values

````python
data = pd.Series([10, 20, 30, np.nan, 50])
print("Range:", data.max() - data.min())
````

## 11. Range and Outliers

Range is highly affected by extreme values.

````python
normal_data = np.array([10, 20, 30, 40, 50])
outlier_data = np.array([10, 20, 30, 40, 500])

print("Normal Range:", np.ptp(normal_data))
print("Range with Outlier:", np.ptp(outlier_data))
````

## 12. Compare Two Datasets

````python
dataset_a = np.array([40, 45, 50, 55, 60])
dataset_b = np.array([10, 30, 50, 70, 90])

print("Dataset A Range:", np.ptp(dataset_a))
print("Dataset B Range:", np.ptp(dataset_b))
````

## 13. Visualization

````python
import matplotlib.pyplot as plt

data = [10, 20, 30, 40, 50, 60, 70, 80]
plt.plot(data, marker="o")
plt.xlabel("Index")
plt.ylabel("Value")
plt.title(f"Data Range = {max(data) - min(data)}")
plt.show()
````

## 14. Practice Exercise

````python
marks = [45, 56, 67, 78, 89, 90, 76, 65, 54, 88]

print("Minimum:", min(marks))
print("Maximum:", max(marks))
print("Range using Python:", max(marks) - min(marks))
print("Range using NumPy:", np.ptp(marks))
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
````
