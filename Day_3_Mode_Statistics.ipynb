# Day 3 — Mode (Statistics)

A beginner-friendly Jupyter Notebook for learning **Mode** using basic Python, `collections.Counter`, Python `statistics`, Pandas, and SciPy.

## 1. What is Mode?

**Mode** is the value that occurs **most frequently** in a dataset.

Example:

`10, 20, 20, 30, 40`

Here, `20` appears two times.

✅ **Mode = 20**

## 2. Find Mode Manually Using Basic Python

````python
data = [10, 20, 20, 30, 40]

frequency = {}

for value in data:
    if value in frequency:
        frequency[value] += 1
    else:
        frequency[value] = 1

print("Frequency:", frequency)

maximum_frequency = max(frequency.values())

mode = []

for value, count in frequency.items():
    if count == maximum_frequency:
        mode.append(value)

print("Mode:", mode)
print("Maximum Frequency:", maximum_frequency)
````

## 3. Mode Using collections.Counter

````python
from collections import Counter

data = [10, 20, 20, 30, 40]

frequency = Counter(data)

print("Frequency:", frequency)

mode = frequency.most_common(1)

print("Most Common Value:", mode[0][0])
print("Frequency:", mode[0][1])
````

## 4. Mode Using Python statistics Module

````python
import statistics

data = [10, 20, 20, 30, 40]

result = statistics.mode(data)

print("Mode:", result)
````

## 5. Mode Using NumPy and SciPy

````python
import numpy as np
from scipy import stats

data = np.array([10, 20, 20, 30, 40])

result = stats.mode(data, keepdims=True)

print("Mode:", result.mode)
print("Count:", result.count)
````

## 6. Mode Using Pandas

````python
import pandas as pd

data = [10, 20, 20, 30, 40]

series = pd.Series(data)

print("Data:")
print(series)

print("\nMode:")
print(series.mode())
````

## 7. Multiple Modes

A dataset can have more than one mode.

Example:

`10, 10, 20, 20, 30`

Both `10` and `20` occur two times.

This is called a **Bimodal Dataset**.

````python
data = [10, 10, 20, 20, 30]

series = pd.Series(data)

print("Modes:")
print(series.mode())
````

## 8. Multimodal Data

If more than two values have the same highest frequency, the dataset can have multiple modes.

````python
data = [10, 10, 20, 20, 30, 30, 40]

frequency = Counter(data)

maximum_frequency = max(frequency.values())

modes = [
    value for value, count in frequency.items()
    if count == maximum_frequency
]

print("Frequency:", frequency)
print("Modes:", modes)
````

## 9. Mode of Categorical Data

Mode is extremely useful for categorical data such as:

- Gender
- City
- Department
- Product
- Color

````python
cities = [
    "Chennai",
    "Madurai",
    "Chennai",
    "Coimbatore",
    "Chennai",
    "Madurai"
]

city_series = pd.Series(cities)

print("City Mode:")
print(city_series.mode())
````

## 10. Student Marks Example

````python
marks = [
    80, 75, 90, 80, 65,
    70, 80, 75, 90, 80
]

marks_series = pd.Series(marks)

print("Marks:")
print(marks)

print("\nFrequency:")
print(marks_series.value_counts())

print("\nMode:")
print(marks_series.mode())
````

## 11. Frequency Analysis Using value_counts()

````python
data = pd.Series([
    "Python", "Java", "Python",
    "SQL", "Python", "Java"
])

frequency = data.value_counts()

print(frequency)

print("\nMost Frequent Value:")
print(frequency.index[0])

print("\nFrequency:")
print(frequency.iloc[0])
````

## 12. Mode With Missing Values

````python
data = pd.Series([
    10, 20, 20,
    np.nan, 30,
    20, np.nan
])

print("Data:")
print(data)

print("\nMode:")
print(data.mode())
````

## 13. Mode Including Missing Values

By default, Pandas usually ignores missing values when finding the mode.

You can explicitly include them using `dropna=False`.

````python
data = pd.Series([
    10, 20, 20,
    np.nan, np.nan
])

print(data.mode(dropna=False))
````

## 14. Group-wise Mode — Data Science Example

````python
df = pd.DataFrame({
    "Department": [
        "IT", "IT", "IT",
        "HR", "HR", "HR",
        "Sales", "Sales", "Sales"
    ],
    "Skill": [
        "Python", "Python", "SQL",
        "Excel", "Excel", "Python",
        "Python", "SQL", "SQL"
    ]
})

print(df)
````

````python
department_mode = df.groupby("Department")["Skill"].agg(
    lambda x: list(x.mode())
)

print("Mode Skill for Each Department:")
print(department_mode)
````

## 15. Real-World Example — Most Popular Product

````python
products = [
    "Laptop",
    "Mouse",
    "Laptop",
    "Keyboard",
    "Laptop",
    "Mouse",
    "Laptop"
]

product_series = pd.Series(products)

print("Product Frequency:")
print(product_series.value_counts())

print("\nMost Popular Product:")
print(product_series.mode())
````

## 16. Mean vs Median vs Mode

````python
data = [10, 20, 20, 30, 100]

print("Data:", data)
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Mode:", pd.Series(data).mode().tolist())
````

### When Should You Use Each?

| Measure | Best Used For |
|---|---|
| Mean | Numerical data without major outliers |
| Median | Numerical data with outliers/skewed data |
| Mode | Most frequent value, including categorical data |

## 17. Visualization of Frequency

````python
import matplotlib.pyplot as plt

data = pd.Series([
    "Python", "Java", "Python",
    "SQL", "Python", "Java"
])

frequency = data.value_counts()

plt.bar(frequency.index, frequency.values)
plt.xlabel("Programming Language")
plt.ylabel("Frequency")
plt.title("Frequency of Programming Languages")
plt.show()
````
