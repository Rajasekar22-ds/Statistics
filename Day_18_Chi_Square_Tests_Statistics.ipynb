## 1. What is a Chi-Square test?
A chi-square test compares **observed counts** with counts we would expect if a particular assumption were true. It is commonly used with categorical data.

Two common types:
- **Goodness of fit:** Does one categorical variable follow an expected pattern?
- **Test of independence:** Are two categorical variables associated?

The data should be counts, observations should be independent, and expected counts should generally not be too small.

## 2. Goodness-of-fit example
A shop expects customers to choose four flavors equally often. We observe the counts below. The null hypothesis says all four flavors are equally popular.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chisquare

observed = np.array([30, 25, 20, 25])
expected = np.array([25, 25, 25, 25])

chi_stat, p_value = chisquare(f_obs=observed, f_exp=expected)
print('Chi-square statistic:', round(chi_stat, 3))
print('P-value:', round(p_value, 4))
alpha = 0.05
if p_value < alpha:
    print('Decision: Reject the null hypothesis.')
else:
    print('Decision: Do not reject the null hypothesis.')
```

```python
labels = ['Vanilla', 'Chocolate', 'Strawberry', 'Mango']
x = np.arange(len(labels))
plt.bar(x - 0.18, observed, width=0.36, label='Observed')
plt.bar(x + 0.18, expected, width=0.36, label='Expected')
plt.xticks(x, labels)
plt.ylabel('Number of customers')
plt.title('Observed vs Expected Counts')
plt.legend()
plt.show()
```

## 3. Test of independence
Use this when you want to check whether two categorical variables are related. Example: is preferred learning method associated with class group? The table contains counts, not percentages.

```python
from scipy.stats import chi2_contingency

# Rows = class groups; columns = preferred learning method
observed_table = np.array([[20, 15, 10],
                           [15, 20, 20]])
chi_stat, p_value, degrees_freedom, expected_table = chi2_contingency(observed_table)
print('Chi-square statistic:', round(chi_stat, 3))
print('Degrees of freedom:', degrees_freedom)
print('P-value:', round(p_value, 4))
print('Expected counts:\n', expected_table)
if p_value < 0.05:
    print('Evidence of an association between the variables.')
else:
    print('Not enough evidence of an association.')
```

## 4. How to interpret the p-value
- A small p-value (often below 0.05) means the observed difference would be unusual if the null hypothesis were true.
- A large p-value does **not** prove the null hypothesis; it means we do not have enough evidence to reject it.
- Statistical association does not prove causation.

## 5. Common mistakes
- Using raw measurements instead of category counts.
- Using a chi-square test for paired measurements without an appropriate method.
- Ignoring small expected counts. If expected counts are too small, consider combining categories when meaningful or using an exact test where appropriate.
- Saying a result proves something with certainty.
