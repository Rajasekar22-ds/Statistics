## 1. What is simple linear regression?

Simple linear regression models the relationship between one numerical input (`x`) and one numerical output (`y`) with a straight line.

The equation is:

\[
\hat{y} = b_0 + b_1x
\]

- **ŷ (y-hat):** predicted output
- **b₀ (intercept):** predicted output when x = 0
- **b₁ (slope):** predicted change in y for a one-unit increase in x

The model describes a pattern in data. It does not automatically prove that x causes y.

## 2. Import the libraries

We use NumPy and pandas for data, Matplotlib for plots, and scikit-learn to fit the regression model.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("Libraries imported successfully!")
```

## 3. Prepare a small dataset

In this example, the input is hours studied and the target is an exam score.

```python
X = np.array([1, 2, 3, 4, 5, 6, 7, 8]).reshape(-1, 1)
y = np.array([48, 52, 60, 63, 67, 72, 78, 85])

data = pd.DataFrame({
    "hours_studied": X.ravel(),
    "exam_score": y
})
display(data)
```

## 4. Train the linear regression model

The `fit()` method learns a slope and intercept from the example data.

```python
model = LinearRegression()
model.fit(X, y)

slope = model.coef_[0]
intercept = model.intercept_

print(f"Slope (b1): {slope:.3f}")
print(f"Intercept (b0): {intercept:.3f}")
print(f"Equation: predicted score = {intercept:.3f} + {slope:.3f} × hours")
```

### Interpret the coefficients

- The **slope** estimates how much the predicted score changes for one additional study hour.
- The **intercept** is the predicted score at zero study hours. It may not be practically meaningful if zero is outside the relevant data range.

## 5. Plot the regression line

The dots are observed data; the line shows the model's predictions.

```python
y_pred = model.predict(X)

plt.figure(figsize=(8, 5))
plt.scatter(X.ravel(), y, label="Observed scores")
plt.plot(X.ravel(), y_pred, label="Regression line")
plt.title("Simple Linear Regression")
plt.xlabel("Hours studied")
plt.ylabel("Exam score")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 6. Make predictions

After fitting the model, use `predict()` to estimate scores for new input values. Predictions far outside the training range are called **extrapolations** and may be unreliable.

```python
new_hours = np.array([2.5, 5.5, 7.5]).reshape(-1, 1)
new_predictions = model.predict(new_hours)

for hours, score in zip(new_hours.ravel(), new_predictions):
    print(f"{hours:g} hours -> predicted score: {score:.2f}")
```

## 7. Understand residuals

A **residual** is the observed value minus the predicted value:

\[
e_i = y_i - \hat{y}_i
\]

- Positive residual: the prediction was too low.
- Negative residual: the prediction was too high.
- Residual close to zero: prediction was close to the observed value.

```python
residuals = y - y_pred

results = pd.DataFrame({
    "hours_studied": X.ravel(),
    "actual_score": y,
    "predicted_score": np.round(y_pred, 2),
    "residual": np.round(residuals, 2)
})
display(results)

plt.figure(figsize=(8, 5))
plt.axhline(0, linestyle="--")
plt.scatter(y_pred, residuals)
plt.title("Residual Plot")
plt.xlabel("Predicted score")
plt.ylabel("Residual (actual - predicted)")
plt.grid(True, alpha=0.3)
plt.show()
```

A residual plot helps you check whether a straight-line model may be missing a pattern. Ideally, residuals are scattered around zero without a clear curve or funnel shape.

## 8. R-squared and error metrics

- **R-squared (R²):** describes how much target variation is accounted for by the model relative to a mean-only baseline under the usual definition.
- **MAE:** average absolute error, in the target's units.
- **MSE:** average squared error, in squared target units.
- **RMSE:** square root of MSE, back in the target's units.

A high R² does not prove causation and does not guarantee good performance on new data. Evaluate a model on held-out data to estimate how well it generalizes.

```python
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y, y_pred)

print(f"MAE:       {mae:.3f}")
print(f"MSE:       {mse:.3f}")
print(f"RMSE:      {rmse:.3f}")
print(f"R-squared: {r2:.3f}")
```
