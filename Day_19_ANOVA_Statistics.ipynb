# Day 19 — ANOVA (Analysis of Variance)

A beginner-friendly guide to comparing means across three or more independent groups.

**How to use:** Run each cell from top to bottom. Read the explanation before running the code.

## 1. What is ANOVA?
**ANOVA** means Analysis of Variance. It tests whether the means of **three or more groups** differ more than we would expect from random variation.

- Null hypothesis (H₀): all population group means are equal.
- Alternative hypothesis (H₁): at least one group mean differs.
- A small p-value is evidence against H₀, but does not tell you which groups differ.

## 2. One-way ANOVA example
Imagine comparing test scores from three teaching methods. Each list contains scores from a different independent group.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import f_oneway

method_A = np.array([78, 82, 85, 80, 79, 84])
method_B = np.array([88, 90, 85, 91, 87, 89])
method_C = np.array([75, 78, 72, 77, 74, 76])

f_stat, p_value = f_oneway(method_A, method_B, method_C)
print('F-statistic:', round(f_stat, 3))
print('P-value:', round(p_value, 6))
if p_value < 0.05:
    print('Reject H0: evidence that at least one group mean differs.')
else:
    print('Do not reject H0: insufficient evidence of a mean difference.')
```

```python
plt.boxplot([method_A, method_B, method_C], labels=['Method A', 'Method B', 'Method C'])
plt.ylabel('Test score')
plt.title('Scores by Teaching Method')
plt.show()
```

## 3. What does the F-statistic mean?
The F-statistic compares variation **between group means** with variation **within groups**. A larger F-statistic often provides more evidence that not all means are equal, but the p-value is used for the decision.

## 4. ANOVA assumptions
1. Observations are independent.
2. Each group's data are approximately normally distributed, especially important with small samples.
3. Group variances are reasonably similar for standard one-way ANOVA.

If variances are clearly unequal, Welch's ANOVA may be more appropriate. If assumptions are seriously violated, consider a suitable alternative such as Kruskal–Wallis.

## 5. What if ANOVA is significant?
ANOVA tells you that at least one mean differs; it does not identify which groups differ. A post-hoc test such as Tukey's HSD can compare pairs while controlling for multiple comparisons.

```python
from scipy.stats import tukey_hsd

result = tukey_hsd(method_A, method_B, method_C)
print(result)
print('Pairwise p-values:\n', result.pvalue)
```
