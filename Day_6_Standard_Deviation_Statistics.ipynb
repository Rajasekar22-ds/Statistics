# Day 6 — Standard Deviation (Statistics)

Beginner-friendly notebook covering Standard Deviation with Python, NumPy, Pandas, and Statistics.

## 1. What is Standard Deviation?

Standard Deviation (SD) measures how far values are spread around the mean.

- Low SD → values are close to the mean.
- High SD → values are more spread out.

**Standard Deviation = √Variance**

## 2. Step-by-Step Calculation

````python
import math

data = [10, 20, 30, 40, 50]

mean = sum(data) / len(data)
squared_differences = [(x - mean) ** 2 for x in data]
variance = sum(squared_differences) / len(data)
standard_deviation = math.sqrt(variance)

print("Data:", data)
print("Mean:", mean)
print("Squared Differences:", squared_differences)
print("Variance:", variance)
print("Population Standard Deviation:", standard_deviation)
````

## 3. Custom Function

````python
def population_standard_deviation(data):
    mean = sum(data) / len(data)
    variance = sum((x - mean) ** 2 for x in data) / len(data)
    return math.sqrt(variance)

print(population_standard_deviation([10, 20, 30, 40, 50]))
````

## 4. NumPy

````python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

print("Population SD:", np.std(data))
print("Sample SD:", np.std(data, ddof=1))
````

## 5. Population vs Sample SD

`ddof=0` gives population SD. `ddof=1` gives sample SD.

## 6. Python Statistics Module

````python
import statistics

data = [10, 20, 30, 40, 50]

print("Population SD:", statistics.pstdev(data))
print("Sample SD:", statistics.stdev(data))
````

## 7. Pandas

````python
import pandas as pd

data = pd.Series([10, 20, 30, 40, 50])

print("Sample SD:", data.std())
print("Population SD:", data.std(ddof=0))
````

## 8. Student Marks Example

````python
marks = np.array([60, 65, 70, 75, 80, 85, 90])

print("Mean:", np.mean(marks))
print("Variance:", np.var(marks))
print("Population SD:", np.std(marks))
print("Sample SD:", np.std(marks, ddof=1))
````

## 9. Multiple Columns

````python
df = pd.DataFrame({
    "Maths": [80, 70, 90, 65, 85],
    "Science": [75, 80, 95, 70, 88],
    "English": [85, 72, 88, 75, 90]
})

print("Sample SD:")
print(df.std())

print("\nPopulation SD:")
print(df.std(ddof=0))
````

## 10. Group-wise Standard Deviation

````python
df = pd.DataFrame({
    "Department": ["IT","IT","IT","HR","HR","HR","Sales","Sales","Sales"],
    "Salary": [30000,35000,45000,25000,28000,30000,32000,36000,40000]
})

print(df.groupby("Department")["Salary"].std())
````

## 11. Missing Values

````python
data = pd.Series([10, 20, 30, np.nan, 50])

print("Sample SD:", data.std())
print("Population SD:", data.std(ddof=0))
````

## 12. Low SD vs High SD

````python
low_spread = np.array([48, 49, 50, 51, 52])
high_spread = np.array([10, 30, 50, 70, 90])

print("Low Spread SD:", np.std(low_spread))
print("High Spread SD:", np.std(high_spread))
````

## 13. Effect of Outliers

````python
normal_data = np.array([10, 20, 30, 40, 50])
outlier_data = np.array([10, 20, 30, 40, 500])

print("Normal SD:", np.std(normal_data))
print("SD With Outlier:", np.std(outlier_data))
````

## 14. Visualization

````python
import matplotlib.pyplot as plt

data = np.array([45, 50, 55, 60, 65, 70, 75, 80])
mean = np.mean(data)
std = np.std(data)

plt.plot(data, marker="o")
plt.axhline(mean, linestyle="--", label=f"Mean = {mean:.2f}")
plt.axhline(mean + std, linestyle=":", label=f"Mean + SD = {mean + std:.2f}")
plt.axhline(mean - std, linestyle=":", label=f"Mean - SD = {mean - std:.2f}")
plt.xlabel("Index")
plt.ylabel("Value")
plt.title("Mean and Standard Deviation")
plt.legend()
plt.show()
````

## 15. Real-World Example — Monthly Sales

````python
sales = pd.DataFrame({
    "Month": ["Jan","Feb","Mar","Apr","May","Jun"],
    "Sales": [100000,120000,110000,150000,90000,130000]
})

print(sales)
print("Mean Sales:", sales["Sales"].mean())
print("Sales SD:", sales["Sales"].std())
````

## 16. Variance and Standard Deviation

````python
data = np.array([10, 20, 30, 40, 50])

variance = np.var(data)
sd = np.std(data)

print("Variance:", variance)
print("Standard Deviation:", sd)
print("Square Root of Variance:", np.sqrt(variance))
````

## 17. Practice Exercise

````python
marks = [45, 56, 67, 78, 89, 90, 76, 65, 54, 88]

print("Mean:", np.mean(marks))
print("Population Variance:", np.var(marks))
print("Population SD:", np.std(marks))
print("Sample SD:", np.std(marks, ddof=1))
print("Pandas Sample SD:", pd.Series(marks).std())
````
