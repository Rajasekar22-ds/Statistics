# Day 2 — Median (Statistics)

Learn Median using basic Python, NumPy, and Pandas.

## 1. What is Median?

Median is the **middle value** after arranging data in ascending order.

Example: `10, 20, 30, 40, 50` → Median = `30`.

## 2. Median with an Even Number of Values

Example: `10, 20, 30, 40`

Median = `(20 + 30) / 2 = 25`.

````python
data = [10, 30, 20, 50, 40]
data.sort()
middle_index = len(data) // 2
median = data[middle_index]
print("Sorted data:", data)
print("Median:", median)
````

````python
data = [40, 10, 30, 20]
data.sort()
n = len(data)
middle1 = data[n // 2 - 1]
middle2 = data[n // 2]
median = (middle1 + middle2) / 2
print("Sorted data:", data)
print("Middle values:", middle1, "and", middle2)
print("Median:", median)
````

## 3. Create Your Own Median Function

````python
def calculate_median(data):
    data = sorted(data)
    n = len(data)
    middle = n // 2

    if n % 2 != 0:
        return data[middle]
    else:
        return (data[middle - 1] + data[middle]) / 2

print(calculate_median([10, 30, 20, 50, 40]))
print(calculate_median([10, 20, 30, 40]))
````

## 4. Median Using NumPy

````python
import numpy as np

data = np.array([10, 30, 20, 50, 40])
print("Median:", np.median(data))
````

## 5. Student Marks Example

````python
marks = np.array([78, 65, 82, 90, 55, 72, 88, 69, 95, 74])

print("Sorted Marks:", np.sort(marks))
print("Median Marks:", np.median(marks))
````

## 6. Row-wise and Column-wise Median

````python
marks = np.array([
    [80, 70, 90],
    [60, 75, 85],
    [90, 88, 95],
    [70, 65, 72],
    [85, 80, 88]
])

print("Overall Median:", np.median(marks))
print("Column-wise Median:", np.median(marks, axis=0))
print("Row-wise Median:", np.median(marks, axis=1))
````

## 7. Median Using Pandas

````python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Kumar", "Arun", "Priya", "Divya"],
    "Marks": [80, 70, 90, 65, 85]
})

print(df)
print("Median Marks:", df["Marks"].median())
````

## 8. Median of Multiple Columns

````python
df = pd.DataFrame({
    "Name": ["Ravi", "Kumar", "Arun", "Priya", "Divya"],
    "Maths": [80, 70, 90, 65, 85],
    "Science": [75, 80, 95, 70, 88],
    "English": [85, 72, 88, 75, 90]
})

print(df.median(numeric_only=True))
````

## 9. Grouped Median

````python
df = pd.DataFrame({
    "Department": ["IT", "IT", "IT", "HR", "HR", "HR", "Sales", "Sales", "Sales"],
    "Salary": [30000, 35000, 100000, 25000, 28000, 30000, 32000, 36000, 38000]
})

print("Overall Median:", df["Salary"].median())
print("\nDepartment-wise Median:")
print(df.groupby("Department")["Salary"].median())
````

## 10. Median With Missing Values

````python
data = [10, 20, 30, np.nan, 50]
df_missing = pd.DataFrame({"Marks": data})

print(df_missing)
print("Median:", df_missing["Marks"].median())
````

## 11. Median and Outliers

Median is less affected by extreme values than Mean.

````python
normal_data = np.array([10, 20, 30, 40, 50])
outlier_data = np.array([10, 20, 30, 40, 500])

print("Normal Data")
print("Mean:", np.mean(normal_data))
print("Median:", np.median(normal_data))

print("\nData With Outlier")
print("Mean:", np.mean(outlier_data))
print("Median:", np.median(outlier_data))
````

## 12. Visualizing Median

````python
import matplotlib.pyplot as plt

data = [10, 20, 30, 40, 50, 60, 70, 80, 90]

plt.boxplot(data)
plt.ylabel("Values")
plt.title("Boxplot Showing the Median")
plt.show()
````

## 14. Practice Exercise

````python
marks = [45, 56, 67, 78, 89, 90, 76, 65, 54, 88]

sorted_marks = sorted(marks)
n = len(sorted_marks)
median_manual = (sorted_marks[n//2 - 1] + sorted_marks[n//2]) / 2

print("Sorted Marks:", sorted_marks)
print("Manual Median:", median_manual)
print("NumPy Median:", np.median(marks))
print("Minimum:", min(marks))
print("Maximum:", max(marks))
````
