```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
np.random.seed(42)
```

## 1. Population, sample, parameter, statistic
- **Population:** the complete group of interest.
- **Sample:** a subset of that group.
- **Parameter:** a value describing the population, such as its true mean.
- **Statistic:** a value calculated from a sample, such as the sample mean.

```python
population = np.array([1, 2, 2, 3, 3, 4, 4, 5, 6, 10])
sample = np.random.choice(population, size=4, replace=False)
print('Population mean:', np.mean(population))
print('Sample:', sample)
print('Sample mean:', np.mean(sample))
```

A sample mean can differ from the population mean. Different samples usually produce different results; this is **sampling variability**.

## 2. Random sampling
A simple random sample gives population members an equal chance of selection. `replace=False` prevents repeats within a sample; `replace=True` allows repeats. Random selection does not automatically remove every possible source of bias.

```python
values = np.array([10,  12, 14, 16, 18, 20])
print('Without replacement:', np.random.choice(values, 3, replace=False))
print('With replacement:', np.random.choice(values, 3, replace=True))
```

## 3. Sampling distribution
A **sampling distribution** shows how a statistic varies across many repeated samples of the same size. We'll sample from a right-skewed population and plot the means.

```python
population_skewed = np.random.exponential(scale=10, size=100_000)
plt.figure(figsize=(8,4))
plt.hist(population_skewed, bins=60, density=True)
plt.title('Original population: right-skewed')
plt.xlabel('Value'); plt.ylabel('Density'); plt.show()

means_n5 = [np.mean(np.random.choice(population_skewed, 5, replace=True)) for _ in range(4000)]
plt.figure(figsize=(8,4))
plt.hist(means_n5, bins=40, density=True)
plt.title('Sampling distribution of means (n=5)')
plt.xlabel('Sample mean'); plt.ylabel('Density'); plt.show()
```

## 4. Central Limit Theorem (CLT)
Under suitable conditions, the distribution of sample means becomes approximately normal as sample size grows, even when the original population is not normal.

Conditions and cautions:
- Samples should be random and observations independent (or dependence handled appropriately).
- The usual CLT assumes finite population variance.
- How large `n` must be depends on the population's shape and tails.
- CLT describes **sample means**, not the raw population data.

```python
fig, axes = plt.subplots(2, 2, figsize=(10,7))
for ax, n in zip(axes.flat, [2, 5, 15, 40]):
    means = [np.mean(np.random.choice(population_skewed, n, replace=True)) for _ in range(3000)]
    ax.hist(means, bins=35, density=True)
    ax.set_title(f'Sample size n={n}')
    ax.set_xlabel('Sample mean')
plt.tight_layout(); plt.show()
```

As `n` grows, sample means often look more bell-shaped and become less spread out.

## 5. Standard error (SE)
Standard deviation describes spread among individual observations. **Standard error** describes how much a statistic varies across repeated samples.

For independent observations with population SD \(\sigma\):

\[SE_{\bar{x}}=\frac{\sigma}{\sqrt{n}}\]

When population SD is unknown, estimate it with sample SD `s`:

\[\widehat{SE}_{\bar{x}}=\frac{s}{\sqrt{n}}\]

```python
sample = np.random.choice(population_skewed, 100, replace=True)
s = np.std(sample, ddof=1)
print('Sample SD:', round(s, 3))
print('Estimated SE:', round(s / np.sqrt(len(sample)), 3))
pop_sd = np.std(population_skewed)
for n in [5, 10, 25, 100, 400]:
    print(f'n={n:3d} | theoretical SE ≈ {pop_sd / np.sqrt(n):.3f}')
```

## 6. CLT vs. Law of Large Numbers (LLN)
- **LLN:** as sample size grows, the sample mean tends to approach the population mean.
- **CLT:** describes the approximate shape of the sample-mean distribution for large samples.

Memory tip: **LLN = gets closer; CLT = shape of the means.**

## 7. Practical example

```python
delivery_minutes = np.random.gamma(shape=4, scale=5, size=50_000)
sample = np.random.choice(delivery_minutes, size=80, replace=False)
print('Sample average:', round(np.mean(sample), 2), 'minutes')
print('Sample SD:', round(np.std(sample, ddof=1), 2), 'minutes')
print('Estimated SE:', round(np.std(sample, ddof=1) / np.sqrt(len(sample)), 2), 'minutes')
```
