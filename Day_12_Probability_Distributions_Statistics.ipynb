# Day 12 — Probability Distributions

A probability distribution tells us how likely the possible values of a random variable are.

**Today:** random variables, discrete vs continuous, PMF, PDF, expected value, variance, standard deviation, Binomial distribution, and Python simulations.

## 1. Random variables

A random variable is a value determined by a random experiment.

Examples:
- Number of Heads in coin tosses
- Number shown by a die
- Number of defective products

**Discrete** variables have countable values (number of students). **Continuous** variables are measurements that can take values in a range (height, weight, time).

```python
import numpy as np
import matplotlib.pyplot as plt

die_values = np.arange(1, 7)
die_probabilities = np.full(6, 1/6)

print("Die values:", die_values)
print("Probabilities:", die_probabilities)
print("Sum of probabilities:", die_probabilities.sum())
```

## 2. Probability distribution and PMF

A probability distribution lists possible values and their probabilities.

A **PMF (Probability Mass Function)** is for discrete random variables. For a fair die, P(X = 4) = 1/6.

Rules: each probability is between 0 and 1, and all probabilities add up to 1.

```python
plt.bar(die_values, die_probabilities)
plt.xlabel("Die value")
plt.ylabel("Probability")
plt.title("PMF of a Fair Die")
plt.xticks(die_values)
plt.show()
```

## 3. PDF — Probability Density Function

A **PDF (Probability Density Function)** describes continuous variables such as height or waiting time. For a continuous variable, we calculate probabilities over an interval (the area under the curve), rather than the probability of one exact value.

| PMF | PDF |
|---|---|
| Discrete variables | Continuous variables |
| Probability of an exact value | Density used to calculate interval probability |
| Number of calls | Waiting time |

## 4. Expected value

Expected value is the long-run average.

**E(X) = Σ x × P(X = x)**

For a fair die, the expected value is (1+2+3+4+5+6)/6 = **3.5**. A die cannot show 3.5; it is an average, not a single outcome.

```python
expected_value = np.sum(die_values * die_probabilities)
print("Expected value of a fair die:", expected_value)

# Example: win ₹100 with probability 0.2; otherwise win ₹0
payouts = np.array([100, 0])
payout_probabilities = np.array([0.2, 0.8])
print("Expected payout: ₹", np.sum(payouts * payout_probabilities))
```

## 5. Variance and standard deviation

Variance measures spread around the mean.

**Var(X) = Σ (x − μ)² × P(X = x)**

Standard deviation is the square root of variance: **SD = √Variance**.

```python
mu = np.sum(die_values * die_probabilities)
variance = np.sum((die_values - mu)**2 * die_probabilities)
std_dev = np.sqrt(variance)

print("Mean:", mu)
print("Variance:", variance)
print("Standard deviation:", std_dev)
```

## 6. Binomial distribution

The **Binomial distribution** models the number of successes in a fixed number of independent trials when each trial has two outcomes and the same probability of success.

Conditions:
1. Fixed number of trials (n)
2. Two outcomes (success/failure)
3. Constant success probability (p)
4. Independent trials

Examples: Heads in coin tosses; defective products in a sample, if assumptions are reasonable.

## 7. Binomial formula

**P(X = x) = C(n, x) × pˣ × (1 − p)ⁿ⁻ˣ**

- n = number of trials
- x = number of successes
- p = probability of success
- C(n, x) = number of ways to choose x successes

Example: probability of exactly 3 Heads in 5 fair coin tosses.

```python
import math

n, x, p = 5, 3, 0.5
probability = math.comb(n, x) * p**x * (1-p)**(n-x)
print("P(exactly 3 Heads in 5 tosses):", probability)
```

## 8. Calculate Binomial probabilities with SciPy

`binom.pmf(x, n, p)` calculates the probability of exactly x successes.

```python
from scipy.stats import binom

n, p = 5, 0.5
x_values = np.arange(n + 1)
binomial_probabilities = binom.pmf(x_values, n, p)

for x, prob in zip(x_values, binomial_probabilities):
    print(f"P(X = {x}) = {prob:.4f}")

print("Sum of probabilities:", binomial_probabilities.sum())

plt.bar(x_values, binomial_probabilities)
plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.title("Binomial Distribution: 5 Fair Coin Tosses")
plt.xticks(x_values)
plt.show()
```

## 9. Binomial mean and variance

For a Binomial distribution:

- **Mean = n × p**
- **Variance = n × p × (1 − p)**
- **Standard deviation = √[n × p × (1 − p)]**

For 5 fair coin tosses, the mean is 2.5 and the variance is 1.25.

```python
n, p = 5, 0.5
print("Mean:", n*p)
print("Variance:", n*p*(1-p))
print("Standard deviation:", np.sqrt(n*p*(1-p)))
```

## 10. Simulate 10,000 experiments

Each experiment contains 5 fair coin tosses. NumPy can simulate the number of Heads in each experiment.

```python
np.random.seed(42)  # Reproducible example
n, p, experiments = 5, 0.5, 10000
results = np.random.binomial(n, p, size=experiments)

print("First 20 results:", results[:20])
print("Simulation mean:", results.mean())
print("Theoretical mean:", n*p)

plt.hist(results, bins=np.arange(-0.5, n+1.5, 1), edgecolor="black")
plt.xlabel("Number of Heads in 5 tosses")
plt.ylabel("Number of experiments")
plt.title("10,000 Simulated Binomial Experiments")
plt.xticks(range(n+1))
plt.show()
```

## 11. Practical example: defective products

Suppose each product has a 2% defect probability. If we inspect 100 independent products, the number of defective products can be modeled with a Binomial distribution.

- n = 100
- p = 0.02
- Expected defects = n × p = 2

```python
n, p = 100, 0.02
print("Expected defects:", n*p)
print("Variance:", n*p*(1-p))
print("Probability of exactly 2 defects:", binom.pmf(2, n, p))
```

## 12. Other distributions

**Discrete:** Bernoulli, Binomial, Poisson.

**Continuous:** Uniform, Normal, Exponential.

### Common mistakes
- PMF is for discrete variables; PDF is for continuous variables.
- Expected value is an average and may not be a possible outcome.
- Standard deviation is the square root of variance.
- Binomial distribution requires a fixed number of independent trials and a constant success probability.
