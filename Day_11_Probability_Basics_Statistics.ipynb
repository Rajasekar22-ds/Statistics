## 1. Basic Idea

Probability is always between **0 and 1**.

**0 ≤ P(A) ≤ 1**

- 0 → impossible
- 0.5 → 50% chance
- 1 → certain

For equally likely outcomes:

**P(A) = Favorable outcomes / Total outcomes**

```python
# Probability of getting 6 on a fair die
favorable = 1
total = 6
p = favorable / total
print("Probability:", p)
print("Percentage:", p * 100, "%")
```

## 2. Random Experiment, Outcome, Sample Space, Event

A **random experiment** has an uncertain result.

Examples: tossing a coin, rolling a die, drawing a card.

An **outcome** is one possible result.

A **sample space** contains all possible outcomes.

Coin: **{H, T}**

Die: **{1, 2, 3, 4, 5, 6}**

An **event** is the outcome or group of outcomes we are interested in.

Example: even numbers = **{2, 4, 6}**

```python
coin_sample_space = {"H", "T"}
die_sample_space = {1, 2, 3, 4, 5, 6}
even_event = {2, 4, 6}

print("Coin:", coin_sample_space)
print("Die:", die_sample_space)
print("Even event:", even_event)
```

## 3. Complement

The complement means the event **does not happen**.

Formula:

**P(Aᶜ) = 1 − P(A)**

If P(rain) = 0.30, then P(no rain) = 0.70.

```python
p_rain = 0.30
print("P(No rain) =", 1 - p_rain)
```

## 4. Addition Rule

If two events cannot happen together:

**P(A or B) = P(A) + P(B)**

Example: probability of getting 1 or 6 on a die = 1/6 + 1/6 = 1/3.

For overlapping events:

**P(A or B) = P(A) + P(B) − P(A and B)**

```python
# 1 or 6 on a die
p_1_or_6 = (1/6) + (1/6)
print("P(1 or 6) =", p_1_or_6)

A = {2, 4, 6}
B = {4, 5, 6}
print("A or B:", A | B)
print("A and B:", A & B)
```

## 5. Independent Events

Independent events do not affect each other.

Example: two coin tosses.

**P(A and B) = P(A) × P(B)**

Two Heads:

1/2 × 1/2 = **1/4 = 25%**

```python
p_two_heads = (1/2) * (1/2)
print("P(Two Heads) =", p_two_heads)
print("Percentage =", p_two_heads * 100, "%")
```

## 6. Dependent Events

Dependent events affect each other.

Example: drawing two Aces **without replacement**.

First Ace = 4/52

Second Ace = 3/51

So:

**P(two Aces) = (4/52) × (3/51)**

```python
p_two_aces = (4/52) * (3/51)
print("P(Two Aces) =", p_two_aces)
```

## 7. Conditional Probability

Conditional probability means the probability of A **given that B has happened**.

**P(A | B) = P(A and B) / P(B)**

Example: 12 students passed and 9 of them studied.

P(Studied | Passed) = 9/12 = **75%**

```python
p_studied_given_passed = 9 / 12
print("P(Studied | Passed) =", p_studied_given_passed)
print("Percentage =", p_studied_given_passed * 100, "%")
```

## 8. Python Coin Simulation

Theoretical probability of Heads = **0.5**.

Let's toss a coin 10 times.

```python
import random

results = [random.choice(["H", "T"]) for _ in range(10)]
heads = results.count("H")
tails = results.count("T")

print("Results:", results)
print("Heads:", heads)
print("Tails:", tails)
print("Experimental P(Heads):", heads / 10)
```

## 9. 10,000 Coin Tosses

With more trials, experimental probability usually gets closer to theoretical probability.

```python
n = 10000
heads = 0

for _ in range(n):
    if random.choice(["H", "T"]) == "H":
        heads += 1

print("Experimental P(Heads):", heads / n)
print("Theoretical P(Heads):", 0.5)
```

## 10. 10,000 Die Rolls

The theoretical probability of getting 6 is **1/6 ≈ 0.1667**.

```python
n = 10000
six_count = 0

for _ in range(n):
    if random.randint(1, 6) == 6:
        six_count += 1

print("Experimental P(6):", six_count / n)
print("Theoretical P(6):", 1/6)
```

```python
import matplotlib.pyplot as plt

rolls = [random.randint(1, 6) for _ in range(10000)]
counts = [rolls.count(i) for i in range(1, 7)]

plt.bar(range(1, 7), counts)
plt.xlabel("Die Number")
plt.ylabel("Frequency")
plt.title("Frequency of 10,000 Die Rolls")
plt.xticks(range(1, 7))
plt.show()
```

## 11. Law of Large Numbers

When we repeat a random experiment many times, the experimental probability tends to get closer to the theoretical probability.

For example:
- 10 coin tosses → result may be far from 0.5
- 10,000 tosses → usually much closer to 0.5

## 12. Practical Example

A factory has a 2% defect probability.

**P(Defective) = 0.02**

**P(Not defective) = 1 − 0.02 = 0.98 = 98%**

```python
p_defective = 0.02
print("P(Defective):", p_defective)
print("P(Not defective):", 1 - p_defective)
```

## 13. Important Rules

| Rule | Formula |
|---|---|
| Probability range | 0 ≤ P(A) ≤ 1 |
| Complement | P(Aᶜ) = 1 − P(A) |
| Mutually exclusive addition | P(A or B) = P(A) + P(B) |
| General addition | P(A or B) = P(A) + P(B) − P(A and B) |
| Independent multiplication | P(A and B) = P(A) × P(B) |
| Conditional probability | P(A\|B) = P(A and B) / P(B) |
