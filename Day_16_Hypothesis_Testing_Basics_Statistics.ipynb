# Day 16 — Hypothesis Testing Basics

Beginner-friendly lesson on null and alternative hypotheses, significance levels, p-values, Type I/II errors, and a one-sample t-test. Run cells from top to bottom. Install SciPy if needed with `pip install scipy`.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
np.random.seed(42)
```

## 1. Hypotheses
**Null hypothesis (H₀):** the starting assumption, often no change or no difference. **Alternative hypothesis (H₁):** the competing claim. Example: H₀: μ = 500 g; H₁: μ ≠ 500 g. Choose one-sided (`>` or `<`) or two-sided (`≠`) alternatives before looking at the data.

## 2. Alpha and errors
The significance level \(\alpha\) is the rejection threshold, often 0.05. **Type I error:** reject a true H₀ (false positive). **Type II error:** fail to reject a false H₀ (false negative). **Power:** chance of rejecting H₀ when a particular alternative is true.

## 3. What is a p-value?
A p-value is the probability, assuming H₀ and the test assumptions are true, of seeing a test statistic at least as extreme as the observed one. It is **not** the probability that H₀ is true, nor the probability that the result happened “by chance.”

## 4. One-sample t-test example
A one-sample t-test compares a sample mean with a hypothesized population mean. It is commonly used when population SD is unknown. Check independence and outliers; with small samples, the data should be reasonably well-behaved.

```python
sample = np.array([503, 498, 501, 497, 505, 499, 502, 496, 504, 500, 501, 498])
t_stat, p_value = stats.ttest_1samp(sample, popmean=500, alternative='two-sided')
alpha = 0.05
print('Sample mean:', round(sample.mean(), 2))
print('t statistic:', round(t_stat, 3))
print('p-value:', round(p_value, 4))
print('Decision:', 'Reject H0' if p_value < alpha else 'Fail to reject H0')
```

## 5. Interpret the result
- If p < α: reject H₀; results are statistically significant at that level.
- If p ≥ α: fail to reject H₀; evidence is insufficient to reject H₀.
Do not say “accept H₀” merely because p ≥ α. A non-significant result does not prove the null is true.

```python
sample = np.array([503, 498, 501, 497, 505, 499, 502, 496, 504, 500, 501, 498])
print('Two-sided:', stats.ttest_1samp(sample, 500, alternative='two-sided'))
print('Greater:', stats.ttest_1samp(sample, 500, alternative='greater'))
print('Less:', stats.ttest_1samp(sample, 500, alternative='less'))
```

## 6. Visualize a two-sided rejection region
The shaded tails show the 5% rejection regions for a t distribution with 11 degrees of freedom.

```python
df, alpha = 11, 0.05
cut = stats.t.ppf(1-alpha/2, df)
x = np.linspace(-4, 4, 500); y = stats.t.pdf(x, df)
plt.figure(figsize=(8,4)); plt.plot(x,y,label='t distribution')
plt.fill_between(x,y,where=(x <= -cut),alpha=.35,label='Rejection regions')
plt.fill_between(x,y,where=(x >= cut),alpha=.35)
plt.axvline(-cut,linestyle='--'); plt.axvline(cut,linestyle='--')
plt.title('Two-sided test: alpha = 0.05'); plt.xlabel('t statistic'); plt.ylabel('Density'); plt.legend(); plt.show()
print('Critical values:', round(-cut,3), round(cut,3))
```

## 7. Statistical vs. practical significance
A tiny effect can be statistically significant with a very large sample. A meaningful effect can be non-significant with a small or noisy sample. Consider effect size, confidence intervals, study design, and real-world context—not only the p-value.
