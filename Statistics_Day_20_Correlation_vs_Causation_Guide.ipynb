## 1. Correlation and causation

**Correlation** means two variables tend to change together.

**Causation** means a change in one variable directly contributes to a change in another.

**Example:** Ice-cream sales and swimming activity may both increase in summer. This does not mean ice cream causes swimming. Warmer weather may affect both.

**Remember:** Correlation alone does not prove causation.

## 3. Pearson correlation coefficient

Pearson's correlation coefficient is written as **r**. It measures the direction and strength of a *linear* relationship between two numerical variables.

| Value of r | Meaning |
|---|---|
| Close to +1 | Strong positive linear relationship |
| Close to -1 | Strong negative linear relationship |
| Close to 0 | Little or no linear relationship |

A value near zero does not rule out every possible relationship; the pattern could be nonlinear. Outliers can also affect Pearson correlation.

## 4. Example: study hours and exam scores

Suppose we record how many hours students study and their exam scores. We will calculate Pearson correlation and draw a scatter plot.

```python
hours_studied = np.array([1, 2, 2, 3, 4, 4, 5, 6, 7])
exam_scores = np.array([50, 55, 58, 62, 68, 70, 75, 82, 88])

r = np.corrcoef(hours_studied, exam_scores)[0, 1]
print(f"Pearson correlation (r): {r:.3f}")

plt.figure(figsize=(8, 5))
plt.scatter(hours_studied, exam_scores)
plt.title("Study Hours vs. Exam Scores")
plt.xlabel("Hours studied")
plt.ylabel("Exam score")
plt.grid(True, alpha=0.3)
plt.show()
```

### Interpretation

The correlation is positive, so students who studied more hours tended to have higher scores in this sample. This is an association, not proof that study hours alone caused the score differences. Sleep, prior knowledge, and other factors may also matter.

## 5. Compare different correlation patterns

Run the following example to see positive, negative, and weak linear patterns side by side.

```python
rng = np.random.default_rng(42)
x = np.arange(1, 31)
positive = 2 * x + rng.normal(0, 5, len(x))
negative = -1.5 * x + rng.normal(0, 5, len(x))
weak = rng.normal(0, 10, len(x))

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, y_values, label in zip(
    axes,
    [positive, negative, weak],
    ["Positive", "Negative", "Weak / unclear"]
):
    corr = np.corrcoef(x, y_values)[0, 1]
    ax.scatter(x, y_values)
    ax.set_title(f"{label}\nr = {corr:.2f}")
    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 6. What is a confounding variable?

A **confounding variable** is a third factor that is related to both variables being studied and may help explain their association.

**Example:**
- Ice-cream sales rise in warm weather.
- Swimming activity may rise in warm weather.
- Temperature may influence both.

A chart or correlation coefficient cannot establish cause and effect by itself. Stronger causal conclusions usually need carefully designed experiments or appropriate causal analysis.

## 7. Correlation using pandas

pandas can calculate a correlation matrix for numerical columns.

```python
data = pd.DataFrame({
    "hours_studied": hours_studied,
    "exam_scores": exam_scores
})

display(data)
print("Correlation matrix:")
display(data.corr(numeric_only=True))
```
