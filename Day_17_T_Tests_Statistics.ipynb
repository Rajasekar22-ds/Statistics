# Day 17 — One-Sample, Independent, and Paired t-Tests

Beginner-friendly Jupyter Notebook with practical Python examples using SciPy, a Welch confidence interval for difference in means, visualization, and practice exercises.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
np.random.seed(42)
```

## 1. Choose the right t-test
- **One-sample t-test:** compare one sample mean with a target.
- **Independent two-sample t-test:** compare two separate groups.
- **Paired t-test:** compare linked observations, such as the same people before and after training.
Always consider independence, outliers, and distribution assumptions.

## 2. One-sample t-test
Example: test whether the average fill amount differs from 100 ml. H₀: μ=100; H₁: μ≠100.

```python
fill_ml = np.array([99,101,98,102,100,97,103,99,100,101,98,102])
t, pval = stats.ttest_1samp(fill_ml, popmean=100)
print('Sample mean:', round(fill_ml.mean(),2), 'ml')
print('t statistic:', round(t,3), 'p-value:', round(pval,4))
```

## 3. Independent two-sample t-test (Welch)
Use for two independent groups. Welch's test (`equal_var=False`) does not assume equal population variances and is a useful default for many comparisons.

```python
group_a = np.array([72,75,71,78,74,73,76,70,77,74])
group_b = np.array([68,70,72,69,71,67,73,70,68,71])
print('Group A mean:', round(group_a.mean(),2))
print('Group B mean:', round(group_b.mean(),2))
t, pval = stats.ttest_ind(group_a, group_b, equal_var=False)
print('Welch t:', round(t,3), 'p-value:', round(pval,4))
print('Decision:', 'Reject H0' if pval < 0.05 else 'Fail to reject H0')
```

## 4. Confidence interval for difference in means
The p-value tests compatibility with H₀; a confidence interval helps show plausible effect sizes. The following calculates a Welch-style 95% interval for mean(A) − mean(B).

```python
from math import sqrt
n1,n2 = len(group_a),len(group_b)
diff = group_a.mean()-group_b.mean()
s1,s2 = np.var(group_a,ddof=1),np.var(group_b,ddof=1)
se = sqrt(s1/n1+s2/n2)
df = (s1/n1+s2/n2)**2 / ((s1/n1)**2/(n1-1)+(s2/n2)**2/(n2-1))
tcrit = stats.t.ppf(0.975,df)
low,high = diff-tcrit*se,diff+tcrit*se
print('Mean difference A-B:', round(diff,2))
print(f'95% CI: ({low:.2f}, {high:.2f})')
```

## 5. Paired t-test
Use when each observation in one condition matches an observation in the other. It tests whether the average within-pair difference is zero.

```python
before = np.array([60,65,70,62,68,72,64,66])
after = np.array([64,67,73,65,70,75,68,69])
print('Average change:', round((after-before).mean(),2))
t,pval = stats.ttest_rel(after,before)
print('Paired t:', round(t,3), 'p-value:', round(pval,4))
```

## 6. Visualize group distributions

```python
plt.figure(figsize=(7,4))
plt.boxplot([group_a,group_b], labels=['Group A','Group B'])
plt.ylabel('Measurement'); plt.title('Two independent groups'); plt.show()
```

## 7. Test selection guide
| Situation | Typical test |
|---|---|
| One sample mean vs. target | One-sample t-test |
| Two independent groups | Welch independent t-test |
| Same people/items measured twice | Paired t-test |

If samples are very small and strongly skewed, or there are influential outliers, standard t-test results may be unreliable; consider alternatives appropriate to the design.
