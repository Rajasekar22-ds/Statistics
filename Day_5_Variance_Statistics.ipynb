# Day 5 — Variance (Statistics)

A beginner-friendly Jupyter Notebook covering **Variance** using Python, NumPy, and Pandas.

## 1. What is Variance?

Variance measures how far the values in a dataset are spread out from the **mean**.

- Small variance → values are close to the mean.
- Large variance → values are widely spread out.

### Population Variance Formula

**σ² = Σ(x − μ)² / N**

### Sample Variance Formula

**s² = Σ(x − x̄)² / (n − 1)**

## 2. Calculate Variance Step by Step

````python
data = [10, 20, 30, 40, 50]

mean = sum(data) / len(data)

squared_differences = []

for value in data:
    difference = value - mean
    squared_difference = difference ** 2
    squared_differences.append(squared_difference)

variance = sum(squared_differences) / len(data)

print("Data:", data)
print("Mean:", mean)
print("Squared Differences:", squared_differences)
print("Population Variance:", variance)
````

## 3. Variance Using a Custom Python Function

````python
def population_variance(data):
    mean = sum(data) / len(data)
    squared_differences = [(x - mean) ** 2 for x in data]
    return sum(squared_differences) / len(data)

data = [10, 20, 30, 40, 50]

print("Population Variance:", population_variance(data))
````

## 4. Variance Using NumPy

````python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

population_variance = np.var(data)

print("Population Variance:", population_variance)
````

## 5. Sample Variance vs Population Variance

NumPy uses `ddof=0` by default for **population variance**.

For **sample variance**, use:

```python
np.var(data, ddof=1)
```

````python
data = np.array([10, 20, 30, 40, 50])

print("Population Variance:", np.var(data))
print("Sample Variance:", np.var(data, ddof=1))
````

## 6. Variance Using Python statistics

````python
import statistics

data = [10, 20, 30, 40, 50]

print("Population Variance:", statistics.pvariance(data))
print("Sample Variance:", statistics.variance(data))
````

## 7. Variance Using Pandas

````python
import pandas as pd

data = pd.Series([10, 20, 30, 40, 50])

print("Sample Variance using Pandas:", data.var())
print("Population Variance:", data.var(ddof=0))
````

## 8. Student Marks Example

````python
marks = np.array([60, 65, 70, 75, 80, 85, 90])

print("Marks:", marks)
print("Mean:", np.mean(marks))
print("Population Variance:", np.var(marks))
print("Sample Variance:", np.var(marks, ddof=1))
````

## 9. Variance of Multiple Columns

````python
df = pd.DataFrame({
    "Maths": [80, 70, 90, 65, 85],
    "Science": [75, 80, 95, 70, 88],
    "English": [85, 72, 88, 75, 90]
})

print("Sample Variance of Each Subject:")
print(df.var())

print("\nPopulation Variance of Each Subject:")
print(df.var(ddof=0))
````

## 10. Group-wise Variance

````python
df = pd.DataFrame({
    "Department": ["IT", "IT", "IT", "HR", "HR", "HR", "Sales", "Sales", "Sales"],
    "Salary": [30000, 35000, 45000, 25000, 28000, 30000, 32000, 36000, 40000]
})

print(df.groupby("Department")["Salary"].var())
````

## 11. Variance With Missing Values

````python
data = pd.Series([10, 20, 30, np.nan, 50])

print("Data:")
print(data)

print("\nSample Variance:")
print(data.var())

print("\nPopulation Variance:")
print(data.var(ddof=0))
````

## 12. Compare Low Variance and High Variance

````python
low_variance = np.array([48, 49, 50, 51, 52])
high_variance = np.array([10, 30, 50, 70, 90])

print("Low Variance Dataset:", np.var(low_variance))
print("High Variance Dataset:", np.var(high_variance))
````

## 13. Variance and Outliers

````python
normal_data = np.array([10, 20, 30, 40, 50])
outlier_data = np.array([10, 20, 30, 40, 500])

print("Normal Data Variance:", np.var(normal_data))
print("Data With Outlier Variance:", np.var(outlier_data))
````

## 14. Visualization

````python
import matplotlib.pyplot as plt

low_variance = [48, 49, 50, 51, 52]
high_variance = [10, 30, 50, 70, 90]

plt.plot(low_variance, marker="o", label="Low Variance")
plt.plot(high_variance, marker="o", label="High Variance")
plt.xlabel("Index")
plt.ylabel("Value")
plt.title("Low Variance vs High Variance")
plt.legend()
plt.show()
````

## 15. Real-World Example — Product Sales

````python
sales = pd.DataFrame({
    "Product": ["A", "B", "C", "D", "E", "F"],
    "Sales": [100, 120, 110, 150, 90, 130]
})

print(sales)

print("\nSales Variance:")
print(sales["Sales"].var())
````

## 16. Practice Exercise

````python
marks = [45, 56, 67, 78, 89, 90, 76, 65, 54, 88]

print("Mean:", np.mean(marks))
print("Population Variance:", np.var(marks))
print("Sample Variance:", np.var(marks, ddof=1))
print("Pandas Sample Variance:", pd.Series(marks).var())
````
