## 1. What is Kurtosis?

In simple words:

> **Kurtosis describes the tail behavior of a distribution.**

The tails are the extreme ends of a distribution.

For example:

```text
                    Center
                      ↓
        ←─────────────●─────────────→
       Left tail                  Right tail
```

Kurtosis is particularly useful when we want to understand how much a dataset may contain **extreme values** compared with a reference distribution.

⚠️ A common beginner mistake is to say that kurtosis simply measures "how peaked" a distribution is. Modern statistical interpretation focuses more on **tail heaviness and extreme observations**.



## 2. Why Do We Need Kurtosis?

Imagine two datasets that have similar means and standard deviations.

One dataset may contain more extreme values than the other.

Kurtosis can help us investigate this difference.

### Simple idea

- More extreme tail observations → higher tail-related kurtosis
- Fewer extreme tail observations → lower tail-related kurtosis

Kurtosis is therefore useful when studying risk, unusual observations, financial returns, errors, and other data where extreme values matter.



## 3. Excess Kurtosis

There are two common ways of reporting kurtosis.

### Pearson kurtosis

For a normal distribution:

**Pearson kurtosis = 3**

### Excess kurtosis

Excess kurtosis subtracts 3:

**Excess kurtosis = Pearson kurtosis − 3**

Therefore, for a normal distribution:

**Excess kurtosis = 0**

This is important because many Python libraries, including SciPy's default `kurtosis()` behavior, report **excess kurtosis**.


````python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import kurtosis

print("Libraries imported successfully!")
````


## 4. Normal Distribution

A normal distribution is commonly used as the reference distribution.

Its excess kurtosis is approximately:

**0**

Its Pearson kurtosis is approximately:

**3**

Let's verify this using a large simulated dataset.


````python

np.random.seed(42)

normal_data = np.random.normal(loc=50, scale=10, size=10000)

excess_kurtosis = kurtosis(normal_data)
pearson_kurtosis = kurtosis(normal_data) + 3

print("Excess kurtosis :", excess_kurtosis)
print("Pearson kurtosis:", pearson_kurtosis)
````

````python

plt.hist(normal_data, bins=50, edgecolor="black")
plt.title("Approximately Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
````


## 5. Three Common Categories

You may see these three terms in statistics:

### 1. Mesokurtic

A distribution with kurtosis similar to the normal distribution.

- Excess kurtosis ≈ 0
- Normal distribution is the standard example

### 2. Leptokurtic

A distribution with relatively heavier tails.

- Excess kurtosis > 0
- More tendency toward extreme observations

### 3. Platykurtic

A distribution with relatively lighter tails.

- Excess kurtosis < 0
- Less tail extremity compared with the normal reference

These terms describe tail behavior relative to a reference distribution.



## 6. Easy Memory Trick

Think about **excess kurtosis**:

| Excess kurtosis | General interpretation |
|---:|---|
| Around 0 | Similar to normal reference |
| Positive | Heavier tails |
| Negative | Lighter tails |

⚠️ These interpretations should be used together with plots and domain knowledge. A numerical kurtosis value alone does not tell the whole story.



## 7. Kurtosis Using SciPy

SciPy provides:

```python
from scipy.stats import kurtosis

kurtosis(data)
```

By default:

- `fisher=True` → returns **excess kurtosis**
- `bias=True` → uses the biased estimator

You can request a bias-corrected estimate with:

```python
kurtosis(data, bias=False)
```


````python

data = [10, 12, 13, 14, 15, 16, 18, 20]

print("Excess kurtosis:", kurtosis(data))
print("Bias-corrected excess kurtosis:", kurtosis(data, bias=False))
````


## 8. Pearson Kurtosis with SciPy

If you want Pearson kurtosis instead of excess kurtosis:

```python
kurtosis(data, fisher=False)
```

Remember:

**Pearson kurtosis = Excess kurtosis + 3**


````python

print("Excess kurtosis :", kurtosis(data))
print("Pearson kurtosis:", kurtosis(data, fisher=False))
````


## 9. Kurtosis Using Pandas

Pandas makes kurtosis calculation easy for Series and DataFrames.

Use:

```python
series.kurt()
```

or:

```python
dataframe.kurt()
```

Pandas reports a form of **excess kurtosis**.


````python

df = pd.DataFrame({
    "Marks": [55, 60, 62, 64, 65, 66, 68, 70, 72, 95],
    "Age": [21, 22, 22, 23, 23, 24, 24, 25, 25, 26]
})

print(df)

print("\nKurtosis:")
print(df.kurt(numeric_only=True))
````


## 10. Creating Heavy-Tailed Data

Let's compare normal data with data generated from a distribution that has heavier tails.

The **Student's t-distribution** is useful for demonstrating heavier tails.


````python

np.random.seed(42)

normal = np.random.normal(0, 1, 10000)
heavy_tailed = np.random.standard_t(df=3, size=10000)

print("Normal excess kurtosis:", kurtosis(normal))
print("Heavy-tailed excess kurtosis:", kurtosis(heavy_tailed))
````

````python

plt.hist(normal, bins=60, edgecolor="black")
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

plt.hist(heavy_tailed, bins=60, edgecolor="black")
plt.title("Heavy-Tailed Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
````


## 11. Outliers and Kurtosis

Extreme observations can have a **large effect on kurtosis**.

Consider:

```text
20, 21, 22, 23, 24, 25
```

Now add:

```text
200
```

The new extreme value can substantially change the tail behavior.

This is one reason kurtosis can be useful when investigating unusual observations.


````python

normal_values = [20, 21, 22, 23, 24, 25]
with_extreme_value = [20, 21, 22, 23, 24, 25, 200]

print("Without extreme value:")
print("Kurtosis:", kurtosis(normal_values))

print("\nWith extreme value:")
print("Kurtosis:", kurtosis(with_extreme_value))
````

````python

plt.boxplot(with_extreme_value)
plt.title("Extreme Value and Boxplot")
plt.ylabel("Value")
plt.show()
````


## 12. Kurtosis vs Skewness

These two concepts are easy to confuse.

### Skewness

Tells us about **asymmetry and direction**.

- Positive → longer right tail
- Negative → longer left tail
- Around 0 → approximately symmetric

### Kurtosis

Tells us about **tail behavior / extreme observations**.

- Positive excess kurtosis → heavier tails
- Negative excess kurtosis → lighter tails
- Around 0 → similar tail behavior to the normal reference

### Easy memory

**Skewness → Which direction?**

**Kurtosis → How heavy are the tails?**



## 13. Mean, Variance, Skewness and Kurtosis

Let's put the major concepts together.

| Statistic | What it tells us |
|---|---|
| Mean | Center of the data |
| Variance | Spread around the mean |
| Standard deviation | Typical spread around the mean |
| Skewness | Asymmetry / direction of the tail |
| Kurtosis | Tail heaviness / extreme observations |

These statistics describe different characteristics of a dataset.



## 14. A Practical Example — Investment Returns

Suppose we have daily investment returns.

Most days may have small movements, but occasionally there may be very large gains or losses.

In this situation, studying kurtosis can help us investigate whether the data has unusually heavy tails.

This does **not** automatically mean that high kurtosis is good or bad.

It simply tells us something about the distribution that may deserve further investigation.


````python

returns = np.array([
    0.5, -0.2, 0.3, -0.1, 0.4,
    0.2, -0.3, 0.1, 0.2, -0.2,
    4.5, -3.8
])

print("Mean:", np.mean(returns))
print("Standard deviation:", np.std(returns))
print("Skewness:", __import__("scipy.stats").stats.skew(returns))
print("Excess kurtosis:", kurtosis(returns))
````


## 15. Visualize the Returns

A histogram can help us see the distribution before interpreting the numerical statistics.


````python

plt.hist(returns, bins=8, edgecolor="black")
plt.axvline(np.mean(returns), linestyle="--", label="Mean")
plt.title("Example of Daily Returns")
plt.xlabel("Return")
plt.ylabel("Frequency")
plt.legend()
plt.show()
````


## 16. A Complete Statistical Profile

For a numerical column, we can calculate several descriptive statistics together.


````python

sales = pd.Series([
    100, 105, 110, 115, 120,
    125, 130, 135, 140, 300
])

summary = {
    "Mean": sales.mean(),
    "Median": sales.median(),
    "Variance": sales.var(),
    "Standard Deviation": sales.std(),
    "Skewness": sales.skew(),
    "Excess Kurtosis": sales.kurt()
}

summary
````


## 17. Practical Data Analysis Workflow

When you receive a new numerical dataset:

### Step 1 — Understand the data
Check what each column represents.

### Step 2 — Calculate descriptive statistics
Mean, median, variance, standard deviation.

### Step 3 — Study the shape
Use histograms and boxplots.

### Step 4 — Check skewness
Understand asymmetry.

### Step 5 — Check kurtosis
Investigate tail behavior and extreme observations.

### Step 6 — Investigate unusual values
Do not automatically delete them.

### Step 7 — Use domain knowledge
An extreme value may be a genuine observation rather than an error.

This is an important data-science habit.



## 18. Important Warning ⚠️

Do not interpret kurtosis as:

> "The bigger the kurtosis, the more peaked the graph."

That explanation is incomplete and can be misleading.

A better beginner interpretation is:

> **Kurtosis, especially excess kurtosis, is mainly useful for understanding tail behavior and the presence of extreme observations relative to a reference distribution.**

Also remember that different software packages can report different definitions of kurtosis. Always check whether the result is **Pearson kurtosis** or **excess kurtosis**.
